### Descripción

Este conjunto de datos contiene información sobre el **Turismo Emisivo** de Uruguay. Incluye información de los residentes uruguayos que viajan al exterior y el gasto realizado según el destino principal del viaje.

Los datos se encuentran disponibles desde el año 2016 e incluyen variables como:

- Transporte internacional de salida e ingreso.
- Fechas de salida y entrada.
- Nacionalidad.
- Departamento de residencia.
- Motivo del viaje.
- Ocupación y nivel educativo del viajero.
- Destino principal.
- Tipo de alojamiento.
- Transporte local.
- Estadía.
- Gastos desagregados en alojamiento, alimentación, transporte, transporte local, cultura, tours, compras, entre otros.

Este conjunto de datos permite analizar el flujo de viajes de los residentes uruguayos al exterior, conocer las principales características de esos viajes y estimar el gasto realizado durante los mismos.

## 7. Valores únicos y su cantidad

Analiza las columnas categóricas para identificar qué valores únicos tienen y la moda. También, verifica cuántos valores únicos tiene cada columna del dataset.

Al observar este dataset, todas las columnas de tipo object corresponden a variables categóricas. Además, existen columnas numéricas que representan esas mismas categorías mediante un código identificador `(Id...)`, por lo que ambas contienen información redundante en distintos formatos. Para analizar los valores únicos y la moda, resulta más útil utilizar las columnas descriptivas de tipo object.

```python
columnas_categoricas = df_emisivo.select_dtypes(include="object").columns # Seleccionamos las columnas de tipo object

for columna in columnas_categoricas: # Recorremos cada columna categórica con un bucle for
    print(f"\n{'='*50}") # \n genera un salto de línea y "=" * 50 crea un separador para mejorar la visualización
    print(f"Columna: {columna}")

    valores_unicos = df_emisivo[columna].unique() # Obtenemos los valores únicos de la columna
    print("\nValores únicos:")
    print(valores_unicos)

    conteo_unicos = df_emisivo[columna].nunique() # Contamos la cantidad de valores únicos
    print(f"\nCantidad de valores únicos: {conteo_unicos}")

    moda = df_emisivo[columna].mode()[0] # Obtenemos la moda de la columna
    print(f"\nModa: {moda}")
```

```python
unicos_por_columna = df_emisivo.nunique().sort_values() # Verificamos la cantidad de valores únicos de todas las columnas del DataFrame y los ordenamos de manera ascendente
print(unicos_por_columna)
```

## 9. Renombrar columnas

Si encuentras columnas con nombres confusos, renómbralas por nombres más claros y adecuados. Explica por qué realizaste los cambios.

```python
for columna in df_emisivo.columns: # Mostramos el nombre de todas las columnas con un bucle for para mejor visualización
    print(columna)
```

Se renombrarán algunas columnas para mejorar la claridad, corregir errores ortográficos y eliminar ambigüedades en los nombres. Los cambios a realizar son los siguientes:

*   **`Pais` a `PaisNacionalidad`**: La columna `Pais` en realidad representa la nacionalidad del viajero, no el país de destino, como se observó en la sección de frecuencia de valores. Se renombra para reflejar correctamente su contenido.
*   **`IdTranspSalidad` a `IdTranspSalida`**: Se corrige un error tipográfico.
*   **`GatoTransporteLocal` a `GastoTransporteLocal`**: Se corrige un error tipográfico.
*   **`TrasporteLocal` a `TransporteLocal`**: Se corrige un error tipográfico.
*   **`Coef` a `GastoCoefExpans`**: El nombre `Coef` es ambiguo. De acuerdo con los metadatos, esta columna representa el 'Coeficiente de expansión del gasto', por lo que se renombra para ser más explícito.
*   **`CoefTot` a `PersonaCoefExpans`**: Similar al anterior, `CoefTot` también es ambiguo. Los metadatos indican que representa el 'Coeficiente de expansión de las personas', por lo que se renombra para mayor claridad.

```python
# Renombramos las columnas utilizando un diccionario, donde cada clave representa
# el nombre actual de la columna y cada valor el nuevo nombre que se le asignará

df_emisivo = df_emisivo.rename(columns={
    "Pais": "PaisNacionalidad",
    "IdTranspSalidad": "IdTransporteSalida",
    "Trasporte Local": "Transporte Local",
    "GatoTransporteLocal": "GastoTransporteLocal",
    "Coef": "CoefExpansionGasto",
    "CoefTot": "CoefExpansionPersonas"
})

for columna in df_emisivo.columns: # # Verificamos que los nuevos nombres de las columnas se hayan aplicado correctamente
    print(columna)
```

##10. Ordenar el dataset

Ordena los datos según una o más columnas de tu elección (numéricas o categóricas). Explica por qué realizaste esta ordenación.

Se ordenará el DataFrame utilizando dos columnas:

*   **`FechaSalida` (ascendente)**: Permite visualizar los registros en orden cronológico, facilitando el análisis de la evolución temporal de los viajes.
*   **`GastoTotal` (descendente)**: Dentro de cada fecha de salida, los viajes con el mayor gasto aparecerán primero, lo que facilita identificar los registros de mayor valor y comparar el comportamiento del gasto entre los viajes realizados en una misma fecha.

```python
# Ordenamos el DataFrame por la fecha de salida en forma ascendente y,
# dentro de cada fecha, por el gasto total en forma descendente

df_emisivo_ordenado = df_emisivo.sort_values(
    by=["FechaSalida", "GastoTotal"],
    ascending=[True, False]
    )

df_emisivo_ordenado.head() # Mostramos las primeras filas del DataFrame ordenado
```

##11. Filtrado de datos

Crea al menos un filtro lógico para trabajar con un subconjunto de datos basado en condiciones específicas que consideres relevantes.

Se creará un subconjunto de datos para los viajes con las siguientes características:

*   **`Destino`: "Argentina"**: Argentina es el destino más frecuente para los viajeros uruguayos, por lo que analizar este segmento puede revelar patrones específicos de gasto y comportamiento.
*   **`Motivo`: "Ocio, Recreo, Vacaciones"**: Este es el motivo de viaje más común. Al combinarlo con el destino Argentina, podemos enfocar el análisis en los viajes de placer a este país, que representan una parte significativa del turismo emisivo.

```python
# Filtramos los viajes con destino Argentina cuyo motivo es ocio, recreo o vacaciones
df_argentina_ocio = df_emisivo_ordenado[
    (df_emisivo_ordenado["Destino"] == "Argentina") &
    (df_emisivo_ordenado["Motivo"] == "Ocio, Recreo, Vacaciones")
]

print(f"Dimensiones del DataFrame filtrado: {df_argentina_ocio.shape}") # Mostramos las dimensiones del DataFrame filtrado

df_argentina_ocio.head() # Mostramos las primeras filas del DataFrame filtrado
```
