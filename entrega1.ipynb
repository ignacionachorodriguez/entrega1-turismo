### Descripción

Este conjunto de datos contiene información sobre el **Turismo Emisivo** de Uruguay. Incluye información de los residentes uruguayos que viajan al exterior y el gasto realizado según el destino principal del viaje.

Los datos se encuentran disponibles desde el año 2016 hasta el año 2025 e incluyen variables como:

- Transporte internacional de salida e ingreso.
- Fechas de salida y entrada.
- Nacionalidad.
- Departamento de residencia.
- Motivo del viaje.
- Ocupación y nivel educativo del viajero.
- Destino principal.
- Tipo de alojamiento.
- Transporte local.
- Estadía.
- Gastos desagregados en alojamiento, alimentación, transporte, transporte local, cultura, tours, compras, entre otros.

Este conjunto de datos permite analizar el flujo de viajes de los residentes uruguayos al exterior, conocer las principales características de esos viajes y estimar el gasto realizado durante los mismos.

Si encuentras columnas con nombres confusos, renómbralas por nombres más claros y adecuados. Explica por qué realizaste los cambios.

```python
for columna in df_emisivo.columns: # Mostramos el nombre de todas las columnas con un bucle for para mejor visualización
    print(columna)
```

Se renombrarán algunas columnas para mejorar la claridad, corregir errores ortográficos y eliminar ambigüedades en los nombres. Los cambios a realizar son los siguientes:

*   **`Pais` a `PaisNacionalidad`**: La columna `Pais` en realidad representa la nacionalidad del viajero, no el país de destino, como se observó en la sección de frecuencia de valores. Se renombra para reflejar correctamente su contenido.
*   **`IdTranspSalidad` a `IdTranspSalida`**: Se corrige un error tipográfico.
*   **`Trasporte Local` a `Transporte Local`**: Se corrige un error tipográfico.
*   **`GatoTransporteLocal` a `GastoTransporteLocal`**: Se corrige un error tipográfico.
*   **`Coef` a `CoefExpansionGasto`**: El nombre `Coef` es ambiguo. De acuerdo con los metadatos, esta columna representa el 'Coeficiente de expansión del gasto', por lo que se renombra para ser más explícito.
*   **`CoefTot` a `CoefExpansionPersonas`**: Similar al anterior, `CoefTot` también es ambiguo. Los metadatos indican que representa el 'Coeficiente de expansión de las personas', por lo que se renombra para mayor claridad.

**Observación:** En una etapa posterior de limpieza y estandarización del conjunto de datos, podría considerarse un renombrado más amplio de las columnas para mantener una convención uniforme en todo el DataFrame. Por ejemplo, unificar el uso de abreviaturas y nombres completos (como `IDTranspSalida` y `Transporte Internacional de Salida`), así como adoptar un criterio consistente respecto al uso de espacios entre palabras.

```python
# Renombramos las columnas utilizando un diccionario, donde cada clave representa
# el nombre actual de la columna y cada valor el nuevo nombre que se le asignará

df_emisivo = df_emisivo.rename(columns={
    "Pais": "PaisNacionalidad",
    "IdTranspSalidad": "IdTranspSalida",
    "Trasporte Local": "Transporte Local",
    "GatoTransporteLocal": "GastoTransporteLocal",
    "Coef": "CoefExpansionGasto",
    "CoefTot": "CoefExpansionPersonas"
})

for columna in df_emisivo.columns: # Verificamos que los nuevos nombres de las columnas se hayan aplicado correctamente
    print(columna)
```

##10. Ordenar el dataset

Ordena los datos según una o más columnas de tu elección (numéricas o categóricas). Explica por qué realizaste esta ordenación.

Se ordenará el DataFrame utilizando dos columnas:

*   **`FechaSalida` (ascendente)**: Permite visualizar los registros en orden cronológico, facilitando el análisis de la evolución temporal de los viajes.
*   **`GastoTotal` (descendente)**: Dentro de cada fecha de salida, los viajes con el mayor gasto aparecerán primero, lo que facilita identificar los registros de mayor valor y comparar el comportamiento del gasto entre los viajes realizados en una misma fecha.

```python
# Ordenamos el DataFrame por la fecha de salida en forma ascendente y,
# dentro de cada fecha, por el gasto total en forma descendente

df_emisivo_ordenado = df_emisivo.sort_values(
    by=["FechaSalida", "GastoTotal"],
    ascending=[True, False]
    )

df_emisivo_ordenado.head() # Mostramos las primeras filas del DataFrame ordenado
```

##11. Filtrado de datos

Crea al menos un filtro lógico para trabajar con un subconjunto de datos basado en condiciones específicas que consideres relevantes.

Se creará un subconjunto de datos para los viajes con las siguientes características:

*   **`Destino`: "Argentina"**: Argentina es el destino más frecuente para los viajeros uruguayos, por lo que analizar este segmento puede revelar patrones específicos de gasto y comportamiento.

##12. Agrupaciones y agregados

Agrupa los datos según alguna columna categórica y calcula al menos una métrica agregada sobre las columnas numéricas (como promedio, suma, etc.).

```python
gasto_promedio_destino = (
    df_emisivo
    .groupby("Destino")["GastoTotal"] # Agrupamos los registros según el destino del viaje
    .mean() # Calculamos el gasto total promedio para cada destino
    .sort_values(ascending=False) # Ordenamos los resultados de mayor a menor
).round(2)

gasto_promedio_destino
```

La documentación del conjunto de datos no especifica si GastoTotal corresponde al gasto del grupo de viaje o al gasto por persona. Por lo tanto, los resultados se interpretan como el gasto total promedio registrado para cada destino, sin asumir que representan un gasto promedio por viajero.
Sin embargo, con la siguiente exploración:

```python
df_emisivo.groupby("Gente")["GastoTotal"].agg(["count", "mean", "median"]).round(2) # Agrupamos por gente, contamos la cantidad de valores, y calculamos el promedio y la mediana
```

Es sugerente que GastoTotal sea el gasto del grupo, no el gasto por persona, ya que se observa que tiende a incrementar cuando aumenta el tamaño del grupo.

##13. Distribución de las columnas numéricas

Analiza la distribución de al menos una columna numérica, describiendo cómo se comportan los valores (puedes identificar tendencias, sesgos, etc.).

```python
import matplotlib.pyplot as plt

df_emisivo["Estadia"].hist(bins=30)

plt.xlim(0, 60)  # Mostramos únicamente las estadías de hasta 60 días

plt.title("Distribución de la duración de la estadía")
plt.xlabel("Días de estadía")
plt.ylabel("Cantidad de viajes")

plt.show()
```

La variable `Estadia` (que representa la cantidad de días de duración de un viaje) presenta una duración promedio de 6,44 días, mientras que la mediana es de 5 días, lo que indica que la mitad de los viajes tuvieron una duración igual o inferior a ese valor. Además, el 75 % de las estadías no supera los 8 días, mientras que la estadía máxima registrada es de 282 días.

```python
porcentaje_estadia = (
    df_emisivo["Estadia"].value_counts() / len(df_emisivo) * 100 # Contamos la frecuencia de los valores de estadía, y obtenemos el porcentaje de cada valor sobre el total
).round(2)

porcentaje_estadia
```

```python
porcentaje_2_a_6 = porcentaje_estadia.loc[[2, 3, 4, 5, 6]].sum() # Sumamos los porcentajes de las estadías cuya duración es entre 2 y 6 días

print(f"El {porcentaje_2_a_6}% de los viajes tienen duración entre 2 y 6 días")
```

Podemos observar que la duración de **estadía** más frecuente es de 3 días, y representa un 14,58 % del total. Además, las **estadías** de entre 2 y 6 días concentran el 61,81 % de los viajes, lo que indica que la mayoría presenta una duración relativamente corta.
